```python
import pandas as pd
import  matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv(r"C:\Users\USER\Downloads\pakistani_ecommerce_dataset.csv")
print(df.head())
print(df.shape)
```

```python
# Q1
print(df.groupby("product_category").count())
print("Most popular category:",df["product_category"].mode()[0])
```

```python
# Q2
print("Average order value:",df["order_value"].mean())
```

```python
#q3
print(df.groupby("city").count())
print("City with most orders:",df["city"].mode()[0])
```

```python
# Q4
print(df.groupby('payment_method').count())
print('Most used payment method:', df['payment_method'].mode()[0])
```

```python
#Q5
plt.figure(figsize=(10,6))
sns.barplot(data=df,x="city",y="order_value",estimator="mean")
plt.xlabel("City")
plt.ylabel("Average Order Value ")
plt.title("Average order value by City")
plt.show()
```
